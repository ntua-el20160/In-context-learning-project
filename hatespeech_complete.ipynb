```python
def create_structured_prompt(dataset, test_example, num_demonstrations=3, dems=None):
    if dems is None:
        demonstrations = random.sample(list(dataset['train']), num_demonstrations)
    else:
        demonstrations = dems

    prompt = ''
    for demo in demonstrations:
        prompt += "Determine whether the following Input is contains hate speech. Label has to be 1 if it contains hate speech and 0 if it does not.\n"
        prompt += f"Input: \"{demo['text']}\"\n"
        prompt += f"Label: {demo['label']}\n"

    prompt += "Determine whether the following Input is contains hate speech. Label has to be 1 if it contains hate speech and 0 if it does not.\n"
    prompt += f"Input: \"{test_example['text']}\"\n"
    prompt += "Label:"

    return prompt
```

```python
import random

def create_prompt_levels(dataset, test_example,rand= False,positive_samples=None,negative_samples=None):


    if positive_samples == None:
      demonstrations = random.sample(list(dataset['train']), 10)
    else:
      pos_sampled = random.sample(positive_samples, 5)
      neg_sampled = random.sample(negative_samples, 5)
      demonstrations = pos_sampled + neg_sampled

    demonstrations = [demo for demo in demonstrations if demo['text'] != test_example['text']]
    random.shuffle(demonstrations)

    f = "Determine whether the following Input is contains hate speech. Label has to be 1 if it contains hate speech and 0 if it does not.\n"
    prompt = ''
    prompts = []
    prompts.append(prompt)
    for count,demo in enumerate(demonstrations[:9]):
        prompt += f
        prompt += f"Input: \"{demo['text']}\"\n"
        if rand:
          label =random.randint(0, 1)
        else:
          label = demo['label']
        prompt += f"Reasoning: {demo['reasoning']}\n"
        prompt += f"Label: {label}\n"

        if(count == 2 or count == 5):
          prompts.append(prompt)
    prompts.append(prompt)
    for i in range(len(prompts)):
      prompts[i] += f
      prompts[i] += f"Input: \"{test_example['text']}\"\n"
      prompts[i] += "Label:"

    return prompts

```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
import torch
from collections import defaultdict

def evaluate_model(dataset,rand = False,dataset_demo = None):
  device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

  prompt_types = {0: "0 demonstrations", 1: "3 demonstrations", 2: "6 demonstrations", 3: "9 demonstrations"}
  #positive_samples = [item for item in dataset['train'] if item['label'] == 1]
  #negative_samples = [item for item in dataset['train'] if item['label'] == 0]
  positive_samples = [item for item in dataset_demo if item['label'] == '1']
  negative_samples = [item for item in dataset_demo if item['label'] == '0']
  results = defaultdict(lambda: {'total_predictions': 0, 'invalid_predictions': 0, 'y_true': [], 'y_pred': []})
  random.seed(42)
  test_set = random.sample(list(dataset['train'])[850:], 1000)
  #test_set = list(dataset['test'])

  for test in test_set:
    prompts = create_prompt_levels(dataset,test,rand,positive_samples,negative_samples)
    for i,prompt in enumerate(prompts):
      #print(prompt)
      #print('\n')
      inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=1024).to(device)

      outputs = model.generate(
        inputs.input_ids,
        max_new_tokens=2,
        temperature=1,
        num_return_sequences=1,
        do_sample=True
      )
      generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
      lines = generated_text.splitlines()

      last_line = ""
      for line in reversed(lines):
          if line.startswith("Label:"):
            last_line = line
            break

      last_line = last_line[6:].strip()
      if last_line == "1" or last_line == "hate":
        results[i]['y_pred'].append(1)
        results[i]['y_true'].append(test['label'])
      elif  last_line == "0":
        results[i]['y_pred'].append(0)
        results[i]['y_true'].append(test['label'])
      else:
         results[i]['invalid_predictions'] +=1

      results[i]['total_predictions'] += 1


  for index, result in results.items():
    invalid_percentage = (result['invalid_predictions'] / result['total_predictions'] * 100) if result['total_predictions'] > 0 else 0

    precision = precision_score(result['y_true'], result['y_pred'], average='weighted', zero_division=0)
    recall = recall_score(result['y_true'], result['y_pred'], average='weighted', zero_division=0)
    f1 = f1_score(result['y_true'], result['y_pred'], average='weighted', zero_division=0)
    accuracy = accuracy_score(result['y_true'], result['y_pred'])

    prompt_label = prompt_types.get(index, f"{index*3} demonstrations")
    print(f"Results for {prompt_label}:")
    print(f"Percentage of invalid predictions: {invalid_percentage:.2f}%")
    print(f"Precision: {precision:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1-score: {f1:.2f}")
    print(f"Accuracy: {accuracy:.2f}")
    print('\n')


```
